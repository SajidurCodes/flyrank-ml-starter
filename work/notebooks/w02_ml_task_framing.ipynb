## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

**Chosen Lane:** Refresh / Content Opportunity Scoring

**ML Task Type:** Classification with ranking.

The objective is to identify pages that are likely to require review for content refresh. The model predicts the probability that a page belongs to the declining category. These probabilities are then used to rank pages so reviewers can prioritize which pages to inspect first.

Although the final output is a ranked list, the underlying machine learning task is a binary classification problem because each page is classified as declining or not declining.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

The primary success metric for this lane is **Precision@K**.

Review teams usually have limited capacity and can only inspect a fixed number of pages. Precision@K measures how many of the top-ranked pages actually deserve review.

Other useful metrics include:

- ROC-AUC
- Average Precision
- Recall

For content prioritization, Precision@K is the most practical because it evaluates the quality of the highest-ranked recommendations.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule evaluates pages using predefined thresholds such as minimum impressions or content age.

Machine learning can learn relationships among many observable signals simultaneously, including impressions, clicks, CTR, engagement, freshness, content length, and search position.

Instead of relying on a single threshold, the model estimates the probability that a page deserves review. This enables better prioritization while still allowing human reviewers to make the final decision.

The model therefore supports human decision-making rather than replacing it.
